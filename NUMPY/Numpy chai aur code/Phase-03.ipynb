## ADVANCE OPERATION WITH BUSINESS EXAMPLES

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Data structure : [restaurant_id, 2021, 2022, 2023, 2024]
sales_data = np.array([
    [1, 150000, 180000, 220000, 250000],  # Paradise biryani
    [2, 120000, 130000, 140000, 160000],  # Beijing Bites
    [3, 200000, 210000, 230000, 260000],  # Pizza Hub
    [4, 100000, 110000, 120000, 130000],  # Burger King
    [5, 170000, 180000, 190000, 200000]   # Taste of India
])

print("==== Zomato Sales Analysis ===")
print("\n Sales Data Shape : ", sales_data.shape)
print("\n Sample data for first three restaurant :\n", sales_data[:3])
```

```python
# Total sales per year
print(np.sum(sales_data, axis=0))

yearly_total = np.sum(sales_data[:, 1:], axis=0)
print(yearly_total)
print(np.sum(yearly_total))

# Minimum sales per restaurant 
min_sales = np.min(sales_data[:, 1:], axis = 1)
print(min_sales)

#Maimum sales per year
max_sales = np.max(sales_data[:, 1:],axis=1)
print(max_sales)

# Average Sales per restaurant
avg_sales = np.mean(sales_data[:, 1:], axis=1)
print(avg_sales)
```

```python
cumlative_sales = np.cumsum(sales_data[:, 1:],axis=1)
print(cumlative_sales)

plt.figure(figsize=(10, 6))
plt.plot(np.mean(cumlative_sales, axis=0))
plt.title("Average Cumulative sales across all restaurant")
plt.ylabel("SALES")
plt.xlabel("Years")
plt.grid(True)
plt.show()
```

```python
vector1 = np.array([1, 2, 3, 4, 5])
vector2 = np.array([6, 7, 8, 9, 10])

print("Vector addition : ", vector1 + vector2)

print("\nMultiplication Vector", vector1 * vector2)

print("\nDot product : ", np.dot(vector1, vector2))

angle = np.arccos(np.dot(vector1, vector2) / (np.linalg.norm(vector1) * np.linalg.norm(vector2)))
print(angle)
```
