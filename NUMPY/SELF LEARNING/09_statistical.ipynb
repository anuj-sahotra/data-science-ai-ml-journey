```python
import numpy as np
import statistics as stats
```

```python
baked_food = [200, 300, 150, 130, 200, 280, 170, 188]

arr = np.array(baked_food)
print(np.mean(arr)) # Mean = Average = sum of all values / total no. of values
print(np.median(arr)) # Median -> First sorted then central value
print(stats.mode(arr)) # Mode - Reccuring value
print(np.std(arr)) #std = Standard deviation
print(np.var(arr)) #var = VARIANCE
print(55.6815 ** 2)
```

```python
tobacco_consumptions = [30, 50, 10, 30, 50, 40]
deaths = [100, 120, 70, 100, 120, 112]

data = np.array([tobacco_consumptions, deaths])

#coefficient of corelation
# -1 - inversely proportional relationship
# 1 - proportinal relationship
# 0 - means no relation
print(np.corrcoef(data))
```
