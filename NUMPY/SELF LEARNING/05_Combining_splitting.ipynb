## Combining And Splitting Array

```python
import numpy as np
```

### List Concatenation

```python
a = [10, 20, 30]
b = [40, 50, 60]
print(a + b)
```

### Array Concatenation

```python
arr1 = np.array([[10, 20], [30, 90]])
arr2 = np.array([[40, 50], [60, 70]])

print(np.concatenate([arr1, arr2]))
```

```python
print(np.concatenate([arr1, arr2], axis=1))  # axis = 1 -> Sideways(row-wise)
```

```python
print(np.concatenate([arr1, arr2], axis=0)) # axis = 0 -> Downwards(column-wise)
```

```python
print(np.hstack([arr1, arr2]))  #hstack() -> Horizontal concatenate
```

```python
print(np.vstack([arr1, arr2]))  #vstack() -> vertical concatenate
```

### ARRAY SPLIT

```python
arr = np.array([20, 30 , 50, 100, 40, 10])
print(np.array_split(arr,4)) # np.array_split(name of array, no. of parts)
```
