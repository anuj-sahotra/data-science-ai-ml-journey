## Adding and removing

```python
import numpy as np
```

```python
arr1 = np.array([10, 20, 30])
print(np.append(arr1, [70, 80]))

arr2 = np.array([[10, 20, 30],
                 [40, 50, 60]])
print(np.append(arr1, arr2))
```

```python
arr = np.array([[10, 20], [30, 40]])
print(np.insert(arr, 1, [50, 60], axis=1)) #array, index, value
```

```python
print(np.delete(arr, 1))
```
