```python
import numpy as np
```

### Sort Array -> np.sort()

```python
arr = np.array([10, 60, 40, 35, 20, 59, 0, 36, 40])
print(np.sort(arr))
```

```python
arr = np.array([[10, 60, 40], [35, 20, 59], [85, 36, 40]])
print(np.sort(arr))
```

# Where -> where(condition)

```python
arr = np.array([[10, 60, 40], [35, 20, 59], [85, 36, 40]])
x = np.where(arr == 35)
print(x)
```

```python
print(np.where(arr % 2 == 0))
```

### Search Sorted -> np.searchsorted(array name, element)

```python
arr = np.array([10, 20, 30, 40, 50])
print(np.searchsorted(arr, 30))
```

```python
arr = np.array([10, 60, 40])
x = arr > 20
print(arr[x])
```
