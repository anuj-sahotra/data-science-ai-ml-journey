## Introduction to numpy

```python
import numpy as np
```

```python
a = np.array([1, 2, 3, 4, 5])
b = np.array([6, 7, 8, 9, 10])
print("Array 1 : ", a)
print("Array 2 : ", b)
```

### Additon and Multplication of an array

```python
print("Array 1 + Array 2 : ", a + b)
print("Array 1 * Array 2 : ", a * b)
```

## Mathematical opeartions

```python
arr1 = np.array([[10, 20], [40, 50]])
arr2 = np.array([[60, 70], [90, 10]])
print(np.add(arr1, arr2))  
print(np.subtract(arr1, arr2))
print(np.divide(arr1, arr2))
print(np.multiply(arr1, arr2))
```

```python
arr1 = np.array([[10, 20], [40, 50]])
arr2 = np.array([2])
print(np.power(arr1, arr2))
```

```python
arr1 = np.array([[100, 225], [400, 625]])
print(np.sqrt(arr1))
```
