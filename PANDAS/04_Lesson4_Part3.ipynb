## Project : Student Performance Data Management

```python
import pandas as pd
```

### Step 1 = Create Dataset

```python
data = {
    "Student" : ["Champion", "Anuj", "Warrior", "Naruto", "Sasuke"],
    "Age" : [23, 20, 25, 33, 34],
    "Study_Hours" : [2, 1, 8, 15, 11],
    "Attendance" : [90, 76, 95, 54, 99],
    "Marks" : [95, 75, 97, 100, 99]
}
df = pd.DataFrame(data)
```

### Step 2 = Save it in CSV

```python
df.to_csv("Student_Performance.csv", index=False)
```

### Step 3 -> Load CSV Again

```python
student_df = pd.read_csv("Student_Performance.csv")
```

### Inspect Dataset

```python
print(f"First 5 rows : \n{student_df.head()}\n")
print(f"Last 5 rows : \n{student_df.tail()}\n")
print(student_df.describe())
```

### Step 5 -> Filter students having marks > 75

```python
high_performers = student_df[
    student_df["Marks"] > 75
]
print(high_performers)
```

### Step 6 - > Export filtered data

```python
high_performers.to_excel("high_performers.xlsx", index=False)
```
