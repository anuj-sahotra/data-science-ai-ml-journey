```python
import pandas as pd
```

### Creaing CSV File

```python
data = {
    "Name" : ["Champion", "Anuj", "Warrior", "Naruto", "Nobita"],
    "Age" : [23, 20, 25, 33, 35],
    "Marks" : [95, 97, 90, 100, 54],
    "City" : ["Kangra", "Nurpur", "Shimla", "Leaf Village", "Tokyo"]
}
df = pd.DataFrame(data)
df.to_csv("Student.csv", index=False)

#index = False -> Pandas default row index ko csv file mein separate column ke roop mein save nhi karega.
```

### Read file

```python
df = pd.read_csv("Student.csv")  # pd.read_csv -> CSV file ko read krta hai
print(df)
```

### Reading only selected columns

```python
df = pd.read_csv(
    "Student.csv",
    usecols=["Name", "City"]
)

print(df)
```

### Reading Limited Rows

```python
df = pd.read_csv("Student.csv", nrows=3)
print(df)
```

### Understanding Important read_csv() Parameters

```python
#index_col= : Make index any column
df = pd.read_csv(
    "Student.csv",
    index_col="Name"
)
print(df)
```

```python
# sep : Column separator specify krna -> CSV mein comma default separator banyega

df = pd.read_csv(
    "Student.csv",
    sep=","
)
```

```python
# dtype : Data Types specify karna

df = pd.read_csv(
    "Student.csv",
    dtype={
        "Age" : "int64",
        "Marks" : "float64"
    }
)
print(df)
print(df.dtypes)
```
