### First practical Pandas program

```python
marks = [90, 80, 70, 60, 50]
student_marks = pd.Series(marks)  # Pandas Series is a 1-D labelled data struture.
print(student_marks)
```

### Basic operation from Series

```python
print(student_marks.sum())  # sum() = Sum of all the values in the Series
print(student_marks.mean()) # mean() = Mean of all the values in the Series
print(student_marks.max())  # max() = Maximum value from the series
print(student_marks.min())  # min() = Minimum Value 
```

```python
"""Task: Ek Pandas Series create karo jisme 5 students ke marks hon.

Requirements:
1. Pandas import karo.
2. Marks ki list banao: [78, 85, 92, 67, 88]
3. List ko Series mein convert karo.
4. Series print karo.
5. Total marks calculate karo.
6. Average marks calculate karo.
7. Highest marks find karo."""

import pandas as pd
marks = [78, 85, 92, 67, 88]
student_marks = pd.Series(marks)
print(student_marks)

print(f"Total Marks : {student_marks.sum()}")
print(f"Average Marks : {student_marks.mean()}")
print(f"Highest Marks : {student_marks.max()}")
```
