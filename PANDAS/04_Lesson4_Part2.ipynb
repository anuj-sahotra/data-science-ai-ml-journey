## PART -2 Excel File

```python
import pandas as pd
```

```python
df = pd.DataFrame({
    "Name" : ["Anuj", "Champion", "Warrior"],
    "Age" : [20, 23, 25],
    "City" : ["Delhi", "Mumbai", "Kolkata"]
})
df.to_excel("Students.xlsx", index=False)
```

```python
df = pd.read_excel("Students.xlsx")
print(df)
```

## Important Excel Parameter

- sheet_name=0 -> First Worksheet
- sheet_name="Sheet1" -> Named Worksheet
- usecols -> Selected Columns
- nrows -> Limited Rows

```python
df = pd.read_excel("Students.xlsx",
                   sheet_name=0)
print(df.keys())
```

```python
df = pd.read_excel("Students.xlsx",
                   sheet_name=None)
print(df.keys())
```
