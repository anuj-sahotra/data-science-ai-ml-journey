## Lesson 3

```python
import pandas as pd
import numpy as np
```

```python
student = {
    "Name" : ["Champion", "Anuj", "Warrior"],
    "Age" : [23, 20, 21],
    "Course" : ["AI/ML", "Cybersecurity", "Data Science"],
    "City" : ["Shimla", "Kangra", "Chamba"]
}

#DataFrame() -> 2D Structure, Row and column labels
df = pd.DataFrame(student) 
print(df)
```

## Creating DataFrame using List 

```python
data = [
    ["Anuj", 23, "Kangra"],
    ["Champion", 20, "Shimla"],
    ["Naruto", 25, "Leaf Village"]
]
# Outer list -> Complete data set
# Inner list -> Individuals rows
# columns -> Column names

df = pd.DataFrame(data,
                  columns=["Name", "Age", "City"])
print(df)
```

## Creating DataFrame using numpy

```python
data = np.array([
    ["Anuj", 23, "Kangra"],
    ["Champion", 20, "Shimla"],
    ["Naruto", 25, "Leaf Village"]
])
df = pd.DataFrame(data,
                  columns = ["Name", "Age", "City"])
print(df)
```

## Creating DataFrame with custom index

```python
data = {
    "Name" : ["Champion", "Anuj", "Warrior"],
    "Age" : [23, 20, 21],
    "Course" : ["AI/ML", "Cybersecurity", "Data Science"],
    "City" : ["Shimla", "Kangra", "Chamba"]
}
df = pd.DataFrame(data,
                  index = ["S101", "S102", "S103"])
print(df)
```

```python
print("Shape : ", df.shape)
print("Size: ", df.size)
print("Dimension : ", df.ndim)
print("Columns : ", df.columns)
print("Index : ", df.index)
print("Data Types : ")
print(df.dtypes)
```

## Lesson 3 - Mini Project

```python
import pandas as pd

# Creating student dataset

data = {
    "Student" : ["Naruto", "Sasuke", "Minato", "Itachi", "Hinato"],
    "Age" : [16, 17, 40, 22, 16],
    "Rank" : [1, 2, 3, 4, 5],
    "Clan" : [["Uzumaki", "Namikaze"], "Uchiha", "Namikaze", "Uchiha", "Hyuga"]
}

# Creating DataFrame
df = pd.DataFrame(data)

#Display dataset
print(df)

# Dataset information
print("\n Dataset Shape : ", df.shape)
print("\n Dataset Size : ", df.size)
print("\n Dataset Columns : ", df.columns)
print("\n Dataset Index : ", df.index)
print("\n Dataset Dimension : ", df.ndim)
print("\n Dataset Data Types : ", df.dtypes)
print("\n Statistical Summary : ", df.describe())
```
