```python
import pandas as pd
```

### Custom Indexing

```python
marks = [98, 97, 78, 99, 54]

students = pd.Series(
    marks,
    index = ["Rahul", "Aman", "Sakshi", "Champion", "Sam"]
)
print(students)
```

### Important attributes of Series

```python
marks = pd.Series([100, 54, 78, 90, 76])

print(marks.index)   # index labels
print(marks.values)  # Actual values
print(marks.dtype)   # Data Type
print(marks.shape)   # Dimension and size
print(marks.size)    # Total elements
print(marks.ndim)    # Number of dimensions
```

### Accessing Series element

```python
# Using Integer location -> iloc[] -> Position based indexing
marks = pd.Series([100, 54, 78, 90, 76],
                  index = ["Rahul", "Aman", "Sakshi", "Champion", "Sam"])
print(marks.iloc[0]) # Returns data of 1st position
print(marks.iloc[3]) # Return data of 4th position
```

```python
# Using loc[] -> Labeled based indexing
print(marks.loc["Champion"])
print(marks.loc["Rahul"])
```

### Mathematical operations on Series

```python
marks = pd.Series([100, 54, 78, 90, 76])

print("Total : ", marks.sum())
print("Maximum : ", marks.max())
print("Minimum : ", marks.min())
print("Average : ", marks.mean())
print("Median : ", marks.median())
print("Standard deviation : ", marks.std())
print(marks.count())
```

### Boolean Filtering

```python
marks = pd.Series([100, 54, 78, 90, 76])

result = marks[marks > 70]

print(result)
```

```python
# Multiple conditions
marks = pd.Series([100, 54, 78, 90, 76])
results = marks[(marks >= 85) | (marks <= 75)]
print(result)
```

### Useful Series method

```python
marks = pd.Series([85, 90, 75, 95, 80, 90])

print("Starting 5 values : \n", marks.head())
print("\nLast 5 values : ", marks.tail())
print("\nUnique values : ", marks.unique())
print("\nUnique values ki count : ", marks.nunique())
print("\nHar value kitni baar aayi : ", marks.value_counts())
print("\nStatistical Summary : ", marks.describe())
```

### Missing values in Series

```python
marks = pd.Series([85, 90, None, 75, None])
print(marks)
```

```python
print(marks.isnull())
print(marks.notnull())
print(marks.isnull().sum())
```

```python
print(marks.fillna(0))  # Replace missing value by 0
```

```python
print(marks.dropna())  # Remove missing value
```
